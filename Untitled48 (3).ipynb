```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

df = pd.read_csv("/content/spam.csv",encoding="latin1")
print(df.head())
print(df.shape)
```

```python
df = df[['v1', 'v2']].copy()

df.columns = ['label', 'text']
df.info()
```

```python

# Initial inspection
print("Dataset shape:", df.shape)
df.info()

print("\nDescriptive Statistics:")
print(df.describe(include="all"))

print("\nClass Distribution:")
print(df["label"].value_counts())
print(df["label"].value_counts(normalize=True) * 100)

print("\nObservation:")
print(f"The dataset contains {len(df)} messages.")
print("The target column is 'label' and input column is 'text'.")
print("Class distribution shows whether spam and ham are balanced.")

# Missing values
print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())

df = df.dropna(subset=["text", "label"]).copy()

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

# Duplicates
duplicate_count = df.duplicated().sum()
print("\nDuplicates Before Removal:", duplicate_count)

df = df.drop_duplicates().reset_index(drop=True)

print("Duplicates After Removal:", df.duplicated().sum())

# Outlier analysis
df["text_length"] = df["text"].str.len()

Q1 = df["text_length"].quantile(0.25)
Q3 = df["text_length"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["text_length"] < lower_bound) |
    (df["text_length"] > upper_bound)
]

print("\nLength Outliers:", len(outliers))
print("IQR Upper Bound:", upper_bound)

```

```python

# Feature Engineering

# Feature 1: text_length
# Longer messages may contain promotional or spam content.
# Already created during outlier analysis.

# Feature 2: word_count
# Word count may help distinguish short personal messages
# from longer promotional messages.
df["word_count"] = df["text"].str.split().str.len()

# Feature 3: has_currency_symbol
# Currency symbols may indicate prizes, money offers,
# or promotional messages associated with spam.
df["has_currency_symbol"] = (
    df["text"].str.contains(r"[$£€]", regex=True)
    .astype(int)
)

print(df[
    ["label", "text_length", "word_count",
     "has_currency_symbol"]
].head())
```

```python

# ==========================================
# FIVE KEY DATASET INSIGHTS
# ==========================================

# Convert labels to readable names for analysis
eda_df = df.copy()

eda_df["label"] = eda_df["label"].replace({
    0: "ham",
    1: "spam",
    "0": "ham",
    "1": "spam"
})

# Calculate statistics
counts = eda_df["label"].value_counts()

percentages = (
    eda_df["label"].value_counts(normalize=True) * 100
)

length_means = (
    eda_df.groupby("label")["text_length"].mean()
)

word_means = (
    eda_df.groupby("label")["word_count"].mean()
)

currency_rates = (
    eda_df.groupby("label")["has_currency_symbol"].mean() * 100
)

print("========== FIVE KEY DATASET INSIGHTS ==========")

print(
    f"\n1. Class Distribution: "
    f"Ham = {percentages.get('ham', 0):.2f}%, "
    f"Spam = {percentages.get('spam', 0):.2f}%."
)

print(
    f"\n2. Missing Values: "
    f"{eda_df[['text', 'label']].isnull().sum().sum()} "
    f"missing values remain after cleaning."
)

print(
    f"\n3. Duplicates: "
    f"{eda_df.duplicated(subset=['text', 'label']).sum()} "
    f"duplicate messages remain in the dataset."
)

print(
    f"\n4. Message Length: "
    f"Ham messages average "
    f"{length_means.get('ham', 0):.2f} characters, "
    f"while spam messages average "
    f"{length_means.get('spam', 0):.2f} characters. "
    f"Ham averages {word_means.get('ham', 0):.2f} words "
    f"and spam averages {word_means.get('spam', 0):.2f} words."
)

print(
    f"\n5. Currency Symbols: "
    f"{currency_rates.get('ham', 0):.2f}% of ham "
    f"messages and "
    f"{currency_rates.get('spam', 0):.2f}% of spam "
    f"messages contain currency symbols."
)
```

```python
x = df["text"].astype(str)
y = df["label"].astype(int)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
tokenizer = Tokenizer(num_words=5000)
tokenizer.fit_on_texts(x_train)
```

```python
x_train = tokenizer.texts_to_sequences(x_train)
x_test = tokenizer.texts_to_sequences(x_test)
```

```python
MAX_LEN = 50
X_train_pad=keras.preprocessing.sequence.pad_sequences(x_train,maxlen=MAX_LEN,padding="post",truncating="post")
X_test_pad=keras.preprocessing.sequence.pad_sequences(x_test,maxlen=MAX_LEN,padding="post",truncating="post")
```

```python
X_train_pad.shape
```

```python
X_test_pad.shape
```

```python
print(X_train_pad[0][:50])
```

```python
rnn_model=keras.Sequential([layers.Input(shape=(MAX_LEN,)),
                             layers.Embedding(5000,32),
                             layers.SimpleRNN(32,activation='tanh'),
                             layers.Dense(1,activation='sigmoid')])
rnn_model.summary()
```

```python
rnn_model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
history=rnn_model.fit(X_train_pad,y_train,epochs=5,batch_size=35,validation_split=0.2,verbose=1)
```

```python
loss,accuracy=rnn_model.evaluate(X_test_pad,y_test)
```

```python
plt.plot(history.history['accuracy'],label='training')
plt.plot(history.history['val_accuracy'],label='validation')
```

```python
plt.plot(history.history['loss'],label='training')
plt.plot(history.history['val_loss'],label='validation')
```

In a plain RNN, gradients can become extremely small when propagated backward across many time steps, so earlier parts of a long sequence learn slowly. This makes it difficult for the model to retain information from words near the beginning of long text.
