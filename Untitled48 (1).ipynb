```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

df = pd.read_csv("/content/spam.csv",encoding="latin1")
print(df.head())
print(df.shape)
```

```python
df.isnull().sum()
```

```python
df = df[["v2", "v1"]]
```

```python
df.columns = ["text", "label"]
```

```python
df = df.dropna(subset=["text", "label"])

print(df.head())
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['label']=le.fit_transform(df['label'])
```

```python
x = df["text"].astype(str)
y = df["label"].astype(int)
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
tokenizer = Tokenizer(num_words=5000)
tokenizer.fit_on_texts(x_train)
```

```python
x_train = tokenizer.texts_to_sequences(x_train)
x_test = tokenizer.texts_to_sequences(x_test)
```

```python
MAX_LEN = 50
X_train_pad=keras.preprocessing.sequence.pad_sequences(x_train,maxlen=MAX_LEN,padding="post",truncating="post")
X_test_pad=keras.preprocessing.sequence.pad_sequences(x_test,maxlen=MAX_LEN,padding="post",truncating="post")
```

```python
X_train_pad.shape
```

```python
X_test_pad.shape
```

```python
print(X_train_pad[0][:50])
```

```python
rnn_model=keras.Sequential([layers.Input(shape=(MAX_LEN,)),
                             layers.Embedding(5000,32),
                             layers.SimpleRNN(32,activation='tanh'),
                             layers.Dense(1,activation='sigmoid')])
rnn_model.summary()
```

```python
rnn_model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
history=rnn_model.fit(X_train_pad,y_train,epochs=5,batch_size=35,validation_split=0.2,verbose=1)
```

```python
loss,accuracy=rnn_model.evaluate(X_test_pad,y_test)
```

```python
plt.plot(history.history['accuracy'],label='training')
plt.plot(history.history['val_accuracy'],label='validation')
```

```python
plt.plot(history.history['loss'],label='training')
plt.plot(history.history['val_loss'],label='validation')
```

In a plain RNN, gradients can become extremely small when propagated backward across many time steps, so earlier parts of a long sequence learn slowly. This makes it difficult for the model to retain information from words near the beginning of long text.
